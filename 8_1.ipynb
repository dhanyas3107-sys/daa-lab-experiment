```python
from itertools import permutations

INF = float('inf')


# ---------------------------------------
# Brute Force TSP
# ---------------------------------------

def tsp_brute_force(cost, n):

    cities = list(range(1, n))

    best_cost = INF
    best_path = None

    for perm in permutations(cities):

        path = [0] + list(perm) + [0]

        c = sum(
            cost[path[i]][path[i + 1]]
            for i in range(n)
        )

        if c < best_cost:
            best_cost = c
            best_path = path

    return best_path, best_cost


# ---------------------------------------
# Nearest Neighbour Heuristic
# ---------------------------------------

def nearest_neighbour(cost, n):

    visited = [False] * n

    path = [0]
    visited[0] = True

    current = 0
    total_cost = 0

    for _ in range(n - 1):

        next_city = -1
        min_cost = INF

        for v in range(n):

            if not visited[v] and cost[current][v] < min_cost:

                min_cost = cost[current][v]
                next_city = v

        path.append(next_city)
        visited[next_city] = True

        total_cost += min_cost
        current = next_city

    # Return to starting city
    path.append(0)
    total_cost += cost[current][0]

    return path, total_cost


# ---------------------------------------
# 6-City Road Network
# ---------------------------------------

cost = [
    [INF, 10, 8, 9, 7, 12],
    [10, INF, 10, 5, 6, 8],
    [8, 10, INF, 8, 9, 7],
    [9, 5, 8, INF, 6, 10],
    [7, 6, 9, 6, INF, 9],
    [12, 8, 7, 10, 9, INF]
]

n = 6

cities = ['A', 'B', 'C', 'D', 'E', 'F']


# ---------------------------------------
# Display Cost Matrix
# ---------------------------------------

print("6-City TSP - Road Distance Matrix:\n")

print(f'{"":>4}', end='')

for c in cities:
    print(f'{c:>6}', end='')

print()

for i, row in enumerate(cost):

    print(f'{cities[i]:>4}', end='')

    for value in row:

        if value == INF:
            print(f'{"INF":>6}', end='')
        else:
            print(f'{value:>6}', end='')

    print()


# ---------------------------------------
# Brute Force
# ---------------------------------------

best_path, best_cost = tsp_brute_force(cost, n)


print("\n=== Brute Force ===")

print(
    f'Optimal Tour: '
    f'{" -> ".join(cities[i] for i in best_path)}'
)

print(f'Minimum Cost: {best_cost} km')


# ---------------------------------------
# Nearest Neighbour
# ---------------------------------------

nn_path, nn_cost = nearest_neighbour(cost, n)


print("\n=== Nearest Neighbour ===")

print(
    f'Tour: '
    f'{" -> ".join(cities[i] for i in nn_path)}'
)

print(f'Total Cost: {nn_cost} km')


# ---------------------------------------
# Comparison
# ---------------------------------------

difference = nn_cost - best_cost

print("\n=== Comparison ===")

print(f'Optimal Cost: {best_cost} km')
print(f'NN Cost: {nn_cost} km')
print(f'Extra Distance: {difference} km')

if best_cost != 0:
    percentage = (difference / best_cost) * 100
    print(f'Percentage Above Optimal: {percentage:.2f}%')
```
