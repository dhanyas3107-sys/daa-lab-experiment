```python
# --- Union-Find for Kruskal ---

class UnionFind:

    def __init__(self, n):
        self.parent = list(range(n))
        self.rank = [0] * n

    def find(self, x):
        if self.parent[x] != x:
            self.parent[x] = self.find(self.parent[x])
        return self.parent[x]

    def union(self, x, y):
        rx, ry = self.find(x), self.find(y)

        if rx == ry:
            return False

        if self.rank[rx] < self.rank[ry]:
            rx, ry = ry, rx

        self.parent[ry] = rx

        if self.rank[rx] == self.rank[ry]:
            self.rank[rx] += 1

        return True


def kruskal(n, edges):
    """Kruskal's Minimum Spanning Tree"""

    edges.sort()

    uf = UnionFind(n)

    mst = []
    cost = 0

    for w, u, v in edges:

        if uf.union(u, v):

            mst.append((u, v, w))
            cost += w

        if len(mst) == n - 1:
            break

    return mst, cost


# --- Graph Definition ---

# A = 0, B = 1, C = 2, D = 3, E = 4, F = 5

n = 6

edges = [
    (4, 0, 1),   # A-B
    (2, 0, 2),   # A-C
    (5, 1, 2),   # B-C
    (10, 1, 3),  # B-D
    (3, 2, 4),   # C-E
    (7, 3, 4),   # D-E
    (8, 3, 5),   # D-F
    (6, 4, 5)    # E-F
]

names = ['A', 'B', 'C', 'D', 'E', 'F']


# --- Apply Kruskal ---

mst, cost = kruskal(n, edges[:])


# --- Display Result ---

print("=== Kruskal's MST ===")

for u, v, w in mst:
    print(f"Edge ({names[u]} - {names[v]}) Weight: {w}")

print(f"Total MST Cost: {cost}")
```
