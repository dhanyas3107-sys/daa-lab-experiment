```python
def solve_n_queens_bitmask(n):

    solutions = []
    backtrack_count = [0]

    # All n columns available
    full_mask = (1 << n) - 1

    def backtrack(row, columns, diag1, diag2, board):

        # All queens placed
        if row == n:
            solutions.append(board[:])
            return

        # Find all safe columns
        available = full_mask & ~(columns | diag1 | diag2)

        while available:

            # Get rightmost available column
            position = available & -available

            # Remove it from available positions
            available -= position

            # Convert bit position to column number
            col = position.bit_length() - 1

            board[row] = col

            # Place queen and update masks
            backtrack(
                row + 1,
                columns | position,
                (diag1 | position) << 1,
                (diag2 | position) >> 1,
                board
            )

            board[row] = -1

            backtrack_count[0] += 1

    board = [-1] * n

    backtrack(0, 0, 0, 0, board)

    return solutions, backtrack_count[0]


# --- Compare N = 12 ---

n = 12

solutions, backtracks = solve_n_queens_bitmask(n)

print(f"N={n}")
print(f"Number of solutions: {len(solutions)}")
print(f"Total backtracks: {backtracks}")
```
