```python
import random

comparison_count = 0


def second_min_max_dc(arr, low, high):
    global comparison_count

    # Base case: single element
    if low == high:
        return arr[low], float('inf'), arr[low], float('-inf')

    # Base case: two elements
    if high == low + 1:

        comparison_count += 1

        if arr[low] < arr[high]:
            return arr[low], arr[high], arr[high], arr[low]

        return arr[high], arr[low], arr[low], arr[high]

    # Divide
    mid = (low + high) // 2

    lmin, lsecond_min, lmax, lsecond_max = \
        second_min_max_dc(arr, low, mid)

    rmin, rsecond_min, rmax, rsecond_max = \
        second_min_max_dc(arr, mid + 1, high)

    # -----------------------------
    # Find minimum and second minimum
    # -----------------------------

    comparison_count += 1

    if lmin < rmin:
        overall_min = lmin

        comparison_count += 1
        if lsecond_min < rmin:
            overall_second_min = lsecond_min
        else:
            overall_second_min = rmin

    else:
        overall_min = rmin

        comparison_count += 1
        if rsecond_min < lmin:
            overall_second_min = rsecond_min
        else:
            overall_second_min = lmin

    # -----------------------------
    # Find maximum and second maximum
    # -----------------------------

    comparison_count += 1

    if lmax > rmax:
        overall_max = lmax

        comparison_count += 1
        if lsecond_max > rmax:
            overall_second_max = lsecond_max
        else:
            overall_second_max = rmax

    else:
        overall_max = rmax

        comparison_count += 1
        if rsecond_max > lmax:
            overall_second_max = rsecond_max
        else:
            overall_second_max = lmax

    return (overall_min, overall_second_min,
            overall_max, overall_second_max)


# --- Demonstration ---

arr = [3, 1, 7, 4, 9, 2, 8, 5, 6, 0]

comparison_count = 0

mn, second_mn, mx, second_mx = \
    second_min_max_dc(arr, 0, len(arr) - 1)

print(f"Array: {arr}")
print(f"Minimum: {mn}")
print(f"Second Minimum: {second_mn}")
print(f"Maximum: {mx}")
print(f"Second Maximum: {second_mx}")
print(f"D&C Comparisons: {comparison_count}")


# --- Performance Analysis ---

print(
    f'\n{"Size":>8} '
    f'{"Comparisons":>15}'
)

print('-' * 30)

for size in [10, 100, 1000, 10000]:

    arr = [
        random.randint(1, 100000)
        for _ in range(size)
    ]

    comparison_count = 0

    mn, second_mn, mx, second_mx = \
        second_min_max_dc(arr, 0, len(arr) - 1)

    print(f'{size:>8} {comparison_count:>15}')
```
