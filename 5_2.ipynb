```python
import random

comparison_count = 0


def partition(arr, low, high):

    global comparison_count

    pivot = arr[high]

    i = low

    for j in range(low, high):

        comparison_count += 1

        if arr[j] <= pivot:

            arr[i], arr[j] = arr[j], arr[i]
            i += 1

    arr[i], arr[high] = arr[high], arr[i]

    return i


def quickselect(arr, low, high, k):

    if low <= high:

        pivot_index = partition(arr, low, high)

        # k is 1-based
        order = pivot_index - low + 1

        if order == k:
            return arr[pivot_index]

        elif k < order:
            return quickselect(
                arr,
                low,
                pivot_index - 1,
                k
            )

        else:
            return quickselect(
                arr,
                pivot_index + 1,
                high,
                k - order
            )

    return None


# --- Test Function ---

def test_quickselect(arr, k):

    global comparison_count

    temp = arr[:]

    comparison_count = 0

    result = quickselect(
        temp,
        0,
        len(temp) - 1,
        k
    )

    return result, comparison_count


# --- Main Execution ---

n = 1000

arr = [
    random.randint(1, 10000)
    for _ in range(n)
]

print(f"Array size: {n}")

# k = 1 -> Minimum
result, comparisons = test_quickselect(arr, 1)

print("\n--- k = 1 (Minimum) ---")
print(f"1st Smallest Element: {result}")
print(f"Comparisons: {comparisons}")


# k = n -> Maximum
result, comparisons = test_quickselect(arr, n)

print("\n--- k = n (Maximum) ---")
print(f"{n}th Smallest Element: {result}")
print(f"Comparisons: {comparisons}")


# k = n // 2 -> Median
k = n // 2

result, comparisons = test_quickselect(arr, k)

print("\n--- k = n//2 (Median) ---")
print(f"{k}th Smallest Element: {result}")
print(f"Comparisons: {comparisons}")
```
