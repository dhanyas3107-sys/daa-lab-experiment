```python
import random
from itertools import permutations

INF = float('inf')


# ---------------------------------------
# Calculate Tour Cost
# ---------------------------------------

def tour_cost(path, cost):

    total = 0

    for i in range(len(path) - 1):
        total += cost[path[i]][path[i + 1]]

    return total


# ---------------------------------------
# Brute Force TSP
# ---------------------------------------

def tsp_brute_force(cost, n):

    cities = list(range(1, n))

    best_cost = INF
    best_path = None

    for perm in permutations(cities):

        path = [0] + list(perm) + [0]

        c = tour_cost(path, cost)

        if c < best_cost:

            best_cost = c
            best_path = path

    return best_path, best_cost


# ---------------------------------------
# Nearest Neighbour
# ---------------------------------------

def nearest_neighbour(cost, n):

    visited = [False] * n

    path = [0]

    visited[0] = True

    current = 0

    for _ in range(n - 1):

        next_city = -1
        min_cost = INF

        for v in range(n):

            if not visited[v] and cost[current][v] < min_cost:

                min_cost = cost[current][v]
                next_city = v

        path.append(next_city)

        visited[next_city] = True

        current = next_city

    path.append(0)

    return path, tour_cost(path, cost)


# ---------------------------------------
# 2-Opt Improvement
# ---------------------------------------

def two_opt(path, cost):

    best_path = path[:]
    best_cost = tour_cost(best_path, cost)

    improved = True

    while improved:

        improved = False

        # Do not change first and last city
        for i in range(1, len(best_path) - 2):

            for j in range(i + 1, len(best_path) - 1):

                # Reverse the selected segment
                new_path = (
                    best_path[:i]
                    + best_path[i:j + 1][::-1]
                    + best_path[j + 1:]
                )

                new_cost = tour_cost(
                    new_path,
                    cost
                )

                # Accept improvement
                if new_cost < best_cost:

                    best_path = new_path
                    best_cost = new_cost

                    improved = True

                    break

            if improved:
                break

    return best_path, best_cost


# ---------------------------------------
# Generate 10-City Random Instance
# ---------------------------------------

random.seed(42)

n = 10

# Random coordinates
coordinates = [
    (
        random.randint(0, 100),
        random.randint(0, 100)
    )
    for _ in range(n)
]


# Create symmetric distance matrix
cost = [[0] * n for _ in range(n)]

for i in range(n):

    for j in range(n):

        if i == j:

            cost[i][j] = INF

        else:

            x1, y1 = coordinates[i]
            x2, y2 = coordinates[j]

            distance = (
                (x1 - x2) ** 2
                + (y1 - y2) ** 2
            ) ** 0.5

            cost[i][j] = round(distance, 2)


# ---------------------------------------
# Nearest Neighbour
# ---------------------------------------

nn_path, nn_cost = nearest_neighbour(
    cost,
    n
)


# ---------------------------------------
# 2-Opt
# ---------------------------------------

two_opt_path, two_opt_cost = two_opt(
    nn_path,
    cost
)


# ---------------------------------------
# Brute Force Optimal Solution
# ---------------------------------------

optimal_path, optimal_cost = tsp_brute_force(
    cost,
    n
)


# ---------------------------------------
# Display Results
# ---------------------------------------

print("=== 10-City Random TSP ===")

print("\nCity Coordinates:")

for i, point in enumerate(coordinates):

    print(f"City {i}: {point}")


print("\n=== Nearest Neighbour ===")

print(
    "Tour:",
    " -> ".join(map(str, nn_path))
)

print(f"Cost: {nn_cost:.2f}")


print("\n=== 2-Opt ===")

print(
    "Tour:",
    " -> ".join(map(str, two_opt_path))
)

print(f"Cost: {two_opt_cost:.2f}")


print("\n=== Brute Force Optimal ===")

print(
    "Tour:",
    " -> ".join(map(str, optimal_path))
)

print(f"Cost: {optimal_cost:.2f}")


# ---------------------------------------
# Comparison
# ---------------------------------------

nn_gap = nn_cost - optimal_cost
two_opt_gap = two_opt_cost - optimal_cost

nn_percentage = (
    nn_gap / optimal_cost
) * 100

two_opt_percentage = (
    two_opt_gap / optimal_cost
) * 100


print("\n=== Comparison ===")

print(f"Optimal Cost       : {optimal_cost:.2f}")
print(f"Nearest Neighbour  : {nn_cost:.2f}")
print(f"2-Opt Cost         : {two_opt_cost:.2f}")

print(
    f"\nNN Gap from Optimal: "
    f"{nn_gap:.2f} ({nn_percentage:.2f}%)"
)

print(
    f"2-Opt Gap from Optimal: "
    f"{two_opt_gap:.2f} ({two_opt_percentage:.2f}%)"
)
```
