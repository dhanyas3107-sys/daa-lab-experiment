```python
import time


# ==========================================
# Bottom-Up Matrix Chain Multiplication
# ==========================================

def matrix_chain_order(dims):

    n = len(dims) - 1

    m = [[0] * (n + 1) for _ in range(n + 1)]
    s = [[0] * (n + 1) for _ in range(n + 1)]

    # Chain length
    for l in range(2, n + 1):

        for i in range(1, n - l + 2):

            j = i + l - 1

            m[i][j] = float('inf')

            for k in range(i, j):

                cost = (
                    m[i][k]
                    + m[k + 1][j]
                    + dims[i - 1] * dims[k] * dims[j]
                )

                if cost < m[i][j]:

                    m[i][j] = cost
                    s[i][j] = k

    return m, s


# ==========================================
# Top-Down Memoized Version
# ==========================================

def matrix_chain_memo(dims):

    n = len(dims) - 1

    # Memoization table
    memo = [[None] * (n + 1) for _ in range(n + 1)]

    # Recursive function
    def solve(i, j):

        # Single matrix
        if i == j:
            return 0

        # Already calculated
        if memo[i][j] is not None:
            return memo[i][j]

        memo[i][j] = float('inf')

        # Try every possible split
        for k in range(i, j):

            cost = (
                solve(i, k)
                + solve(k + 1, j)
                + dims[i - 1] * dims[k] * dims[j]
            )

            if cost < memo[i][j]:
                memo[i][j] = cost

        return memo[i][j]

    cost = solve(1, n)

    return cost, memo


# ==========================================
# Main
# ==========================================

# Six matrices
# A1(30x35)
# A2(35x15)
# A3(15x5)
# A4(5x10)
# A5(10x20)
# A6(20x25)

dims = [30, 35, 15, 5, 10, 20, 25]

n = len(dims) - 1

print("Matrix Dimensions:")

for i in range(n):

    print(
        f" A{i + 1}: "
        f"{dims[i]} x {dims[i + 1]}"
    )


# ==========================================
# Bottom-Up DP
# ==========================================

start = time.perf_counter()

m, s = matrix_chain_order(dims)

bottom_up_cost = m[1][n]

bottom_up_time = (
    time.perf_counter() - start
) * 1000


# ==========================================
# Top-Down Memoization
# ==========================================

start = time.perf_counter()

memo_cost, memo = matrix_chain_memo(dims)

memo_time = (
    time.perf_counter() - start
) * 1000


# ==========================================
# Results
# ==========================================

print("\n=== Bottom-Up DP ===")
print(f"Minimum Cost: {bottom_up_cost}")
print(f"Time: {bottom_up_time:.6f} ms")


print("\n=== Top-Down Memoization ===")
print(f"Minimum Cost: {memo_cost}")
print(f"Time: {memo_time:.6f} ms")


print("\n=== Verification ===")

if bottom_up_cost == memo_cost:
    print("Both approaches give the same minimum cost.")
else:
    print("Results do not match.")
```
