```python
import random
import time


# -------------------------------
# Interpolation Search
# for Floating-Point Numbers
# -------------------------------
def interpolation_search_float(arr, target):
    low, high = 0, len(arr) - 1
    comparisons = 0

    while low <= high and arr[low] <= target <= arr[high]:
        comparisons += 1

        if low == high:
            if arr[low] == target:
                return low, comparisons
            return -1, comparisons

        # Avoid division by zero
        if arr[high] == arr[low]:
            if arr[low] == target:
                return low, comparisons
            return -1, comparisons

        # Interpolation formula
        pos = low + int(
            ((target - arr[low]) * (high - low))
            / (arr[high] - arr[low])
        )

        # Make sure position stays within the valid range
        pos = max(low, min(pos, high))

        if arr[pos] == target:
            return pos, comparisons
        elif arr[pos] < target:
            low = pos + 1
        else:
            high = pos - 1

    return -1, comparisons


# -------------------------------
# Performance Analysis
# -------------------------------
def performance_analysis():

    sizes = [10000, 50000, 100000]

    print(
        f"{'Size':>10} "
        f"{'Time(ms)':>15} "
        f"{'Comparisons':>15}"
    )

    print("-" * 45)

    for size in sizes:

        # Generate uniformly distributed floating-point numbers
        arr = sorted(
            random.uniform(0.0, 1000.0)
            for _ in range(size)
        )

        # Select an existing element as target
        target = arr[random.randint(0, size - 1)]

        # Repeat search to get more reliable timing
        repetitions = 100

        start = time.perf_counter()

        for _ in range(repetitions):
            idx, comparisons = interpolation_search_float(
                arr, target
            )

        elapsed_time = (
            (time.perf_counter() - start)
            / repetitions
            * 1000
        )

        print(
            f"{size:>10} "
            f"{elapsed_time:>15.6f} "
            f"{comparisons:>15}"
        )


# -------------------------------
# Main
# -------------------------------

# Small demonstration
arr = [0.5, 1.2, 2.8, 4.5, 6.7, 8.9, 10.5]

target = 6.7

idx, comparisons = interpolation_search_float(
    arr, target
)

print("Array:", arr)
print("Searching for:", target)
print("Found at index:", idx)
print("Comparisons:", comparisons)

print("\nPerformance Analysis")
performance_analysis()
```
