```python
# --- Knight's Tour using Backtracking ---

N = 8

# All possible Knight moves
moves = [
    (2, 1),
    (1, 2),
    (-1, 2),
    (-2, 1),
    (-2, -1),
    (-1, -2),
    (1, -2),
    (2, -1)
]


def is_safe(board, row, col):

    return (
        0 <= row < N and
        0 <= col < N and
        board[row][col] == -1
    )


def count_next_moves(board, row, col):

    count = 0

    for dr, dc in moves:

        new_row = row + dr
        new_col = col + dc

        if is_safe(board, new_row, new_col):
            count += 1

    return count


def solve_knight_tour():

    board = [
        [-1] * N
        for _ in range(N)
    ]

    backtrack_count = [0]

    # Starting position
    board[0][0] = 0

    def backtrack(row, col, move_count):

        # All cells visited
        if move_count == N * N:
            return True

        # Generate possible moves
        candidates = []

        for dr, dc in moves:

            new_row = row + dr
            new_col = col + dc

            if is_safe(board, new_row, new_col):

                degree = count_next_moves(
                    board,
                    new_row,
                    new_col
                )

                candidates.append(
                    (degree, new_row, new_col)
                )

        # Try moves with fewer onward moves first
        candidates.sort()

        for _, new_row, new_col in candidates:

            board[new_row][new_col] = move_count

            if backtrack(
                new_row,
                new_col,
                move_count + 1
            ):
                return True

            # Undo move
            board[new_row][new_col] = -1

            backtrack_count[0] += 1

        return False

    # Start with move number 1
    if backtrack(0, 0, 1):
        return board, backtrack_count[0]

    return None, backtrack_count[0]


def display_board(board):

    print("\nKnight's Tour Solution:\n")

    print("+" + "----+" * N)

    for row in range(N):

        print("|", end="")

        for col in range(N):

            print(
                f"{board[row][col]:2d} |",
                end=""
            )

        print()

        print("+" + "----+" * N)


# --- Main ---

board, backtracks = solve_knight_tour()

if board:

    display_board(board)

    print(f"\nStarting Position: (0, 0)")
    print(f"Board Size: {N} x {N}")
    print(f"Cells Visited: {N * N}")
    print(f"Backtracks: {backtracks}")

else:

    print("No Knight's Tour found.")
```
