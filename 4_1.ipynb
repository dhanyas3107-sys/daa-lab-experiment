```python
import heapq


def dijkstra(graph, source):
    """
    Dijkstra's Algorithm using Min-Heap
    Time: O((V + E) log V)
    """

    n = len(graph)

    dist = [float('inf')] * n
    prev = [None] * n

    dist[source] = 0

    pq = [(0, source)]
    visited = set()

    while pq:

        d, u = heapq.heappop(pq)

        if u in visited:
            continue

        visited.add(u)

        for v, w in graph[u]:

            if dist[u] + w < dist[v]:

                dist[v] = dist[u] + w
                prev[v] = u

                heapq.heappush(pq, (dist[v], v))

    return dist, prev


def reconstruct_path(prev, source, target):

    path = []

    node = target

    while node is not None:
        path.append(node)
        node = prev[node]

    path.reverse()

    if path[0] == source:
        return path

    return []


# --------------------------
# City Road Network
# --------------------------

graph = {
    0: [(1, 4), (2, 2)],   # Airport
    1: [(4, 5)],
    2: [(3, 1), (4, 7)],
    3: [(4, 3)],
    4: []                  # Central Station
}

source = 0
target = 4

dist, prev = dijkstra(graph, source)

path = reconstruct_path(prev, source, target)

print("Airport (0) -> Central Station (4)")
print("Fastest Route:", " -> ".join(map(str, path)))
print("Travel Time:", dist[target], "minutes")
```
