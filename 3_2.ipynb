```python
import heapq
import random
import time


# --------------------------------
# Prim's Algorithm - Adjacency Matrix
# --------------------------------

def prim_matrix(n, graph, start=0):

    INF = float('inf')

    key = [INF] * n
    parent = [-1] * n
    inMST = [False] * n

    key[start] = 0

    mst = []
    cost = 0

    for _ in range(n):

        # Find vertex with minimum key
        min_value = INF
        u = -1

        for v in range(n):

            if not inMST[v] and key[v] < min_value:
                min_value = key[v]
                u = v

        if u == -1:
            break

        inMST[u] = True

        if parent[u] != -1:
            mst.append((parent[u], u, graph[parent[u]][u]))
            cost += graph[parent[u]][u]

        # Update adjacent vertices
        for v in range(n):

            if (graph[u][v] != 0 and
                not inMST[v] and
                graph[u][v] < key[v]):

                key[v] = graph[u][v]
                parent[v] = u

    return mst, cost


# --------------------------------
# Prim's Algorithm - Heap Version
# --------------------------------

def prim_heap(n, adj, start=0):

    INF = float('inf')

    key = [INF] * n
    parent = [-1] * n
    inMST = [False] * n

    key[start] = 0

    pq = [(0, start)]

    mst = []
    cost = 0

    while pq:

        w, u = heapq.heappop(pq)

        if inMST[u]:
            continue

        inMST[u] = True

        if parent[u] != -1:
            mst.append((parent[u], u, w))
            cost += w

        for v, wt in adj.get(u, []):

            if not inMST[v] and wt < key[v]:

                key[v] = wt
                parent[v] = u

                heapq.heappush(pq, (wt, v))

    return mst, cost


# --------------------------------
# Generate Dense Graph
# --------------------------------

def generate_dense_graph(n):

    matrix = [[0] * n for _ in range(n)]
    adj = {}

    for i in range(n):
        adj[i] = []

    for i in range(n):

        for j in range(i + 1, n):

            weight = random.randint(1, 100)

            matrix[i][j] = weight
            matrix[j][i] = weight

            adj[i].append((j, weight))
            adj[j].append((i, weight))

    return matrix, adj


# --------------------------------
# Performance Comparison
# --------------------------------

n = 100

matrix, adj = generate_dense_graph(n)


# Matrix Prim
start = time.perf_counter()

matrix_mst, matrix_cost = prim_matrix(n, matrix)

matrix_time = (time.perf_counter() - start) * 1000


# Heap Prim
start = time.perf_counter()

heap_mst, heap_cost = prim_heap(n, adj)

heap_time = (time.perf_counter() - start) * 1000


# --------------------------------
# Results
# --------------------------------

print("=== Prim's Algorithm Comparison ===")

print(f"Number of vertices: {n}")

print("\n--- Adjacency Matrix ---")
print(f"MST Cost: {matrix_cost}")
print(f"Time: {matrix_time:.4f} ms")
print("Time Complexity: O(V^2)")

print("\n--- Heap + Adjacency List ---")
print(f"MST Cost: {heap_cost}")
print(f"Time: {heap_time:.4f} ms")
print("Time Complexity: O(E log V)")
```
